````python
import os
import subprocess
import sys
import threading
from pathlib import Path

from dotenv import load_dotenv

from make_ca_bundle import build as build_ca_bundle

load_dotenv(override=True)
build_ca_bundle()

PROJECT = Path.cwd()


def show(line):
    encoding = getattr(sys.stdout, "encoding", None) or "utf-8"
    print(line.encode(encoding, "replace").decode(encoding, "replace"), end="")


def run_step(step, timeout=1800):
    process = subprocess.Popen(
        [sys.executable, "-m", "lymphoma", step],
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        cwd=PROJECT,
        text=True,
        encoding="utf-8",
        errors="replace",
        bufsize=1,
    )
    watchdog = threading.Timer(timeout, process.kill)
    watchdog.start()
    try:
        with process:
            for line in process.stdout:
                show(line)
    finally:
        watchdog.cancel()
    if process.returncode != 0:
        raise RuntimeError(f"{step} failed with code {process.returncode}")


print("Python:", sys.executable)
print("Folder:", PROJECT)
print("OPENAI_API_KEY:", bool(os.getenv("OPENAI_API_KEY")))
print("TAVILY_API_KEY:", bool(os.getenv("TAVILY_API_KEY")))
print("CCWAI_API_KEY:", bool(os.getenv("CCWAI_API_KEY") or os.getenv("CURE_CANCER_WITH_AI_API_KEY")))
````
